```python
s = pd.Series([1,2,3], index=["a", "b", "c"], name="sequence") # this is a list, the index is provided as a separate argument.
# name argument gives a name to the Series, which can be useful for identification and labeling purposes.

s.name # this is the name of the Series, which is "sequence" in this case.
```

```python
pd.Series({"a": 1, "b": 2}) # this is a dictionary, the keys become the index and the values become the data
```

```python
arr = np.array([3,5,7])
s = pd.Series(arr, copy=False) 
s[1] = -1
arr
```

```python
s = pd.Series([1,2,3], index=["a", "b", "c"], name="sequence")
s["b"] # direct access using the index label, not recommended.

s.loc["b"] # recommended way to access data using the index label.

s.iloc[1] # used only for integer position based indexing, not recommended for label based access.
# iloc on a label does not work
```

It is recommended to use loc or iloc function

```python
s1 = pd.Series([1,2,3])
s1.iloc[1]
s1.loc[1]
```

loc is for labels that are defined!

```python
s = pd.Series({"a": 1, "b": 2, "c": 3, "d": 4, "e": 5, "f":6})
s.iloc[3::2] # starting point, endpoint, step size
s
```

```python
s.iloc[1:3] # does not include the endpoint
```

```python
s.loc["b":"c"] # does include the endpoint
```

```python
s1 = pd.Series({"a":3, "b":5, "c":2})
s2 = pd.Series({"b":3, "a":5, "c":2})
s1 + s2 # the addition is done based on the index labels, not the order of the data.
```

```python
s1 = pd.Series({"a":3, "b":5, "c":2,"d":1})
s2 = pd.Series({"b":3, "a":5, "c":2})
s1 + s2
```

```python
# 100 values standard normal
rng = np.random.default_rng(12345)
import matplotlib.pyplot as plt
s = pd.Series(rng.standard_normal(100))
# line, hist, box, kde
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10,6))
s.plot(kind='line', ax=axes[0,0]) ; axes[0,0].set_title("Line plot")
s.plot(kind='kde', ax=axes[0,1])  ; axes[0,1].set_title("KDE plot")
s.plot(kind='box', ax=axes[1,0])  ; axes[1,0].set_title("Box plot")
s.plot(kind='hist', ax=axes[1,1]) ; axes[1,1].set_title("Histogram plot")
fig.suptitle("Line, KDE, Box, and Histogram plots of the Series")
plt.show()
```
