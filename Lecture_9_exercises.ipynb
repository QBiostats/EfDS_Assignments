# Grouping

```python
%%sql
SELECT AlbumId, COUNT(TrackId) AS TracksNum, ROUND(SUM(UnitPrice), 2) AS TotalPrice, ROUND(AVG(Milliseconds) / 1000.0, 0) AS MeanTrackSec
    FROM tracks
    GROUP BY AlbumId
    ORDER BY TotalPrice DESC
    LIMIT 10
```

```python
engine = sa.create_engine("sqlite:///chinook.db")
query = "SELECT * FROM tracks"

conn = engine.raw_connection()
try:
    df_tracks = pd.read_sql(query, con=conn)
finally:
    conn.close()

df_sum = df_tracks.groupby('AlbumId').agg(TracksNum=('TrackId', 'count'), TotalPrice=('UnitPrice', 'sum'), MeanTrackSec=('Milliseconds', 'mean')).reset_index()

df_sum['MeanTrackSec'] = (df_sum['MeanTrackSec'] / 1000).round(0)
df_sum['TotalPrice'] = df_sum['TotalPrice'].round(2)
df_sum = df_sum.sort_values(by='TotalPrice', ascending=False).head(10)

df_sum
```

```python
%%sql
SELECT AlbumId, COUNT(TrackId) AS TrackCount
    FROM tracks
    GROUP BY AlbumId
    HAVING TrackCount BETWEEN 18 AND 20
    LIMIT 5
```

```python
%%sql
SELECT AlbumId, GROUP_CONCAT(Name, "; ") AS Tracks
    FROM tracks
    GROUP BY AlbumId
    LIMIT 5
```
