## Unidirected graph (class, writing to file/console, exceptions)

```python
import sys

class Graph:
    def __init__(self): # step 1
        self._vs = set()
        self._es = set()

    def addV(self, vertex): # step 2 and 4
        if vertex in self._vs:
            raise ValueError(f"Vertex {vertex} is already in the graph")
        self._vs.add(vertex)
        return self

    def addE(self, v1, v2): # step 3 and 4
        if v1 not in self._vs:
            raise ValueError(f"Add {v1} first as a vertex")
        if v2 not in self._vs:
            raise ValueError(f"Add {v2} first as a vertex")
    
        edge = tuple(sorted((v1,v2)))

        if edge in self._es:
            raise ValueError(f"Edge {v1} <->{v2} is already in the graph")

        self._es.add(edge)
        return self
    
    def vs(self): # step 5
        return sorted(list(self._vs))

    def es(self):
        return list(self._es)

    def hasE(self, v1, v2): # step 6
        edge = tuple(sorted((v1, v2)))
        return edge in self._es

    def writeAdjacencyMatrix(self, f=sys.stdout): # step 7
        nodes = self.vs()
        f.write("    " + "   ".join(nodes) + "\n")
        
        for row_node in nodes:
            line = [row_node]
            for col_node in nodes:
                if self.hasE(row_node, col_node):
                    line.append("1")
                else:
                    line.append("0")
            f.write("   ".join(line) + "\n")

if __name__ == "__main__":
    g = Graph()
    g.addV("a")
    g.addV("b")
    g.addV("c").addV("d")

    g.addE("a", "b")
    g.addE("c", "c").addE("c", "d")

    print("Vertices:", g.vs())
    print("Edges:   ", g.es())

    print("Check a-b:", g.hasE("a", "b"))
    print("Check a-c:", g.hasE("a", "c"))

    print("\nAdjacency Matrix:")
    g.writeAdjacencyMatrix()

    with open("g.mx", "w") as f:
        g.writeAdjacencyMatrix(f=f)
```

## Fibonacci numbers (functions)

```python
def fib(pos):
    if pos == 0:
        return 0
    if pos == 1:
        return 1
    
    a, b = 0, 1

    for _ in range(2, pos + 1):
        a, b = b, a + b

    return b
```

```python
[fib(pos) for pos in range(19)]
```
