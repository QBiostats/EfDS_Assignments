```python
def identity_(n):
    matrix = np.zeros((n,n), dtype=int)

    for i in range(n):
        matrix[i, i] = 1
    return matrix
```

```python
def grid_x(n):
    grid=np.full((n,n), ".")

    index = np.arange(n)

    grid[index, index] = "#"
    grid[index, index[::-1]] = "#"

    return grid
```

## Array with summary

```python
import numpy as np
```

```python
arr2d = np.random.randint(0,100, size=(8,4))
print(arr2d)
```

```python
miminum = np.min(arr2d, axis =0)
maximum = np.max(arr2d, axis =0)
mean = np.mean(arr2d, axis =0)
median = np.median(arr2d, axis =0)
q1 = np.percentile(arr2d, 25, axis=0)
q3 = np.percentile(arr2d, 75, axis =0)
```

```python
def summary(arr):
    mins = np.min(arr, axis=0)
    q1 = np.percentile(arr, 25, axis=0)
    median = np.median(arr, axis=0)
    mean = np.mean(arr, axis=0)
    q3 = np.percentile(arr, 75, axis=0)
    maxs = np.max(arr, axis=0)

    num_cols = arr.shape[1]

    labels = ["Min.   :", "1st Qu.:", "Median :", "Mean   :", "3rd Qu.:", "Max.   :"]
    stats_matrix = [mins, q1, median, mean, q3, maxs]

    header = ""
    for i in range(num_cols):
        header += f"{i:^20}"
    print(header)

    for row_idx, label in enumerate(labels):
        row_str = ""
        for col_idx in range(num_cols):
            val = stats_matrix[row_idx][col_idx]
            cell_content = f"{label}{val:<7.2f}"
            row_str += f"{cell_content:<20}"
        print(row_str)
```

```python
summary(arr2d)
```

```python
def mat_mult(A, B):
    rows_A, cols_A = A.shape
    rows_B, cols_B = B.shape

    if cols_A != rows_B:
        raise ValueError(
            f"Incompatible dimensions for multiplication: {cols_A} columns in A "
            f"and {rows_B} rows in B."
        )

    result = np.zeros((rows_A, cols_B))
    
    for i in range(rows_A):
        for j in range(cols_B):
            for k in range(cols_A):
                result[i, j] += A[i, k] * B[k, j]
```
