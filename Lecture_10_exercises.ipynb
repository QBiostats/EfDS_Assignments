# Joining basics

```python
%reload_ext sql
%sql sqlite:///chinook.db
```

## Customer invoices

```python
%%sql
SELECT CustomerId, FirstName, LastName, InvoiceId
    FROM customers
    LEFT JOIN invoices USING (CustomerId)
```

CustomerId and InvoiceId are the combined primary key

## Customers without invoices

```python
%%sql
SELECT CustomerId, FirstName, LastName, InvoiceId
    FROM customers
    LEFT JOIN invoices USING (CustomerId)
    WHERE InvoiceId IS NULL
```

## Double join

```python
customer_summary = %sql SELECT \
    c.CustomerId, \
    COUNT(ii.TrackId) AS TotalTracksNum, \
    SUM(ii.UnitPrice) AS TotalPrice \
FROM customers c \
JOIN invoices i ON c.CustomerId = i.CustomerId \
JOIN invoice_items ii ON i.InvoiceId = ii.InvoiceId \
GROUP BY c.CustomerId
```
