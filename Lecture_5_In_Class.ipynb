```python
class WordCounter:
    def __init__(self):
        self.word2count = defaultdict(int)

    def countWordsInText(self, text):
        for word in text.split():
            self.word2count[word] += 1
            
    def getNumOfAllWords(self):
        return sum(self.word2count.values())

    def get_word2count(self):
        return self.word2count
```

```python
def ourDivision(div):
    res = float("nan")
    print("Start")
    try:
        print("Start of try")
        res = 1 / div # this line may crash, depending on the value of "div" argument
        print("End of try")
    except ZeroDivisionError:
        print("I caught ZeroDivisionError")
    except TypeError:
        print("You did not provide a number")
    print("End")
    return res
```

```python
ourDivision(1)
```

## Files

getting access to the file

```python
f = open(file = "Test_5.txt", mode = "r") # mode is what you want to do with the file: "r" for reading, "w" for writing, "a" for appending, etc.
lines = f.readlines()
f.close()
print(lines)
```

Shorter way to get to the file: it reads the whole document at ones

```python
with open(file = "Test_5.txt", mode = "r") as f:
    lines = f.readlines()
print(lines)
```

Reading line by line: a loop

```python
with open(file = "Test_5.txt", mode = "r") as f:
    for line in f:
        print(line)
```

```python
with open(file = "Test_5.txt", mode = "r") as f:
    ret = [(l, len(l)) for l in f]
print(ret)
```

writing names to a file

```python
with open(file = "my_written_names.txt", mode = "w") as writeFileHandle:
    for word in names.split():
        print(word, file = writeFileHandle)
```
